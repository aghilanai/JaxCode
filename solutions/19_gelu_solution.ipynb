import jax.numpy as jnp
def my_gelu(x):
    # tanh approximation (matches jax.nn.gelu default)
    return 0.5 * x * (1.0 + jnp.tanh(jnp.sqrt(2.0 / jnp.pi) * (x + 0.044715 * x**3)))
