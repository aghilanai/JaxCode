# ✅ Solution: Implement Dropout

Implement **Dropout** regularization from scratch.

### Signature
```python
class MyDropout(nnx.Module):
    def __init__(self, p: float = 0.5): ...
    def forward(self, x: Tensor) -> Tensor: ...
```

### Rules
- During **training**: zero each element with probability `p`, scale remaining by `1/(1-p)`
- During **eval**: return input unchanged (identity)
- Do NOT use `nn.Dropout` or `F.dropout`

````python
# ✅ SOLUTION

import jax, jax.numpy as jnp
from flax import nnx
class MyDropout(nnx.Module):
    def __init__(self,p=0.5,*,rngs): self.p=p; self.rngs=rngs
    def __call__(self,x,deterministic=False):
        if deterministic or self.p==0: return x
        return x*(jax.random.uniform(self.rngs.dropout(),x.shape)>=self.p)/(1-self.p)
````
