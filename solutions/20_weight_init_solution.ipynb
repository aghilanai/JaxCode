````python
# ✅ SOLUTION

import jax, jax.numpy as jnp, math
def kaiming_init(shape,key):
    fan_in=shape[-1] if len(shape)>=2 else shape[0]
    return jax.random.normal(key,shape)*math.sqrt(2.0/fan_in)
````
